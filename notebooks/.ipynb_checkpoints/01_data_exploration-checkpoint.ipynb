```python
gender_dict = {
    "Ahmed": "male",
    "Sara": "female",
    "Omar": "male",
    "Mona": "female",
    "Ali": "male",
    "Youseef": "male",
    "nada": "female",
    "Laila": "female"
}
gender_from_name = df_clean["customer_name"].map(gender_dict)
```

```python
df_clean["customer_gender"] = df_clean["customer_gender"].fillna(gender_from_name)
```
